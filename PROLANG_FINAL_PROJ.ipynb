#### A. Data Types

* **Example A.1 — Primitive Types:** [Example A.1](./project-overview/language-analysis/A-Data_Types/example_1-1.ts)
  * **Implementation:**
    ```typescript
    let favNumber: number = 67;
    let longNumber: number = 1337;
    let pi: number = 3.14;
    let bigPi: number = 3.1415926535;

    console.log(favNumber, longNumber, pi, bigPi);
    ```
  * **Analysis:** TypeScript uses the number type for all
    numeric values, integers, floats, and doubles are all
    just number. TypeScript is statically typed
    when type annotations are used meaning the type is
    declared explicitly and the compiler enforces it at
    compile time. Assigning a string to a number variable
    would immediately throw a compile error.

* **Example A.2 — String:** [Example A.2](./project-overview/language-analysis/A-Data_Types/example_1-2.ts)
  * **Implementation:**
    ```typescript
    let name: string = "BOB";
    let greeting: string = `Hello ${name}`;

    console.log(greeting);
    ```
  * **Analysis:** string is a primitive type in TypeScript.
    TypeScript string primitives are immutable
    value-like types. TypeScript also supports template
    literals using backticks which allow embedded
    expressions directly inside the string using ${}
    syntax.

* **Example A.3 — Nullable / Undefined:** [Example A.3](./project-overview/language-analysis/A-Data_Types/example_1-3.ts)
  * **Implementation:**
    ```typescript
    let nullableAge: number | null = null;
    let undefinedValue: string | undefined = undefined;

    console.log(nullableAge);       // null
    console.log(undefinedValue);    // undefined
    ```
  * **Analysis:** TypeScript handles nullable types using
    union types with | null or | undefined.
    TypeScript uses union types to explicitly allow null
    or undefined as valid values. This is part of
    TypeScript's strict null checking system which when
    enabled forces you to handle null and undefined
    cases explicitly before using a value.

* **Example A.4 — Type Inference:** [Example A.4](./project-overview/language-analysis/A-Data_Types/example_1-4.ts)
  * **Implementation:**
    ```typescript
    let score = 67;
    let playerName = "John Kaisen";

    console.log(`${playerName} scored ${score} points`);
    ```
  * **Analysis:** TypeScript can infer the type
    automatically from the assigned value without an
    explicit annotation. score becomes number and
    playerName becomes string at compile time.

* **Example A.5 — Array and Tuple:** [Example A.5](./project-overview/language-analysis/A-Data_Types/example_1-5.ts)
  * **Implementation:**
    ```typescript
    let scores: number[] = [85, 90, 78, 92];
    let player: [string, number] = ["John Kaisen", 67];

    console.log(scores[0]);     // 85
    console.log(player[0]);     // John Kaisen
    console.log(player[1]);     // 67
    ```
  * **Analysis:** Arrays in TypeScript are typed using
    the type[] syntax. TypeScript also has tuple types
    which are fixed length arrays where each position
    has a specific type. A tuple [string, number] means
    index 0 must always be a string and index 1 must
    always be a number, the compiler enforces this.

* **Example A.6 — Boolean:** [Example A.6](./project-overview/language-analysis/A-Data_Types/example_1-6.ts)
  * **Implementation:**
    ```typescript
    let isAlive: boolean = true;
    let hasKey: boolean = false;

    console.log(`John Kaisen alive: ${isAlive}, Has key: ${hasKey}`);
    ```
  * **Analysis:** boolean is a primitive type in TypeScript
    that holds either true or false. It is statically
    typed so assigning a number or string to a boolean
    variable throws a compile error. TypeScript is
    stricter than JavaScript here, in JavaScript any
    value can be truthy or falsy, but in TypeScript a
    boolean must explicitly be true or false.

* **Example A.7 — Any and Unknown:** [Example A.7](./project-overview/language-analysis/A-Data_Types/example_1-7.ts)
  * **Implementation:**
    ```typescript
    let flexible: any = 42;
    flexible = "now a string";
    flexible = true;

    let safe: unknown = "hello";
    if (typeof safe === "string") {
        console.log(safe.toUpperCase());
    }
    ```
  * **Analysis:** any disables type checking entirely,
    a variable of type any can hold any value and
    TypeScript will not check it. This is unique to
    TypeScript. unknown is a safer alternative, it also accepts any value
    but forces you to check the type before using it.
    any should be avoided in production code as it
    defeats the purpose of using TypeScript. unknown
    is the preferred choice when the type is genuinely
    not known ahead of time.

#### B. Expressions and Assignment Statements

* **Example B.1 — Arithmetic:** [Example B.1](./project-overview/language-analysis/B-Expressions_Assignment-Statements/example_2-1.ts)
  * **Implementation:**
    ```typescript
    let a: number = 10;
    let b: number = 3;

    console.log(a + b);   // Addition — 13
    console.log(a - b);   // Subtraction — 7
    console.log(a * b);   // Multiplication — 30
    console.log(a / b);   // Division — 3.333...
    console.log(a % b);   // Modulus — 1

    a++;  // Increment — 11
    b--;  // Decrement — 2
    console.log(a);
    console.log(b);
    ```
  * **Analysis:** TypeScript supports the standard
    arithmetic operators. TypeScript always returns a floating point result,
    10 / 3 returns 3.3333 not 3. This is because
    TypeScript only has one number type for all
    numerics. The ++ and -- operators work by incrementing and decrementing by 1.

* **Example B.2 — Compound Assignment:** [Example B.2](./project-overview/language-analysis/B-Expressions_Assignment-Statements/example_2-2.ts)
  * **Implementation:**
    ```typescript
    let x: number = 5;

    x += 3;   // x = 8
    x -= 2;   // x = 6
    x *= 4;   // x = 24
    x /= 6;   // x = 4
    x %= 3;   // x = 1

    console.log(x);
    ```
  * **Analysis:** Compound assignment operators combine
    an arithmetic operation with assignment in one step.
    They are commonly used inside loops to
    update a running value without having to write the
    full expression each time.

* **Example B.3 — Comparison Operators:** [Example B.3](./project-overview/language-analysis/B-Expressions_Assignment-Statements/example_2-3.ts)
  * **Implementation:**
    ```typescript
    let score: number = 80;

    console.log(score == 80);   // true -  equal to
    console.log(score != 100);  // true -  not equal to
    console.log(score > 75);    // true -  greater than
    console.log(score < 75);    // false - less than
    console.log(score >= 80);   // true -  greater than or equal to
    console.log(score <= 79);   // false - less than or equal to
    ```
  * **Analysis:** TypeScript uses === for strict equality
    which checks both value and type. This is different
    from JavaScript's == which only checks value and
    performs type coercion. In TypeScript === is always
    preferred over == because it prevents unexpected
    comparisons like 0 == false being true.

* **Example B.4 — Boolean Logic:** [Example B.4](./project-overview/language-analysis/B-Expressions_Assignment-Statements/example_2-4.ts)
  * **Implementation:**
    ```typescript
    let hasKey: boolean = true;
    let doorOpen: boolean = false;

    console.log(hasKey && !doorOpen);   // AND — true
    console.log(hasKey || doorOpen);    // OR — true
    console.log(!hasKey);               // NOT — false
    ```
  * **Analysis:** Boolean logic operators. && returns true only if both
    sides are true. || returns true if at least one
    side is true. ! flips the value. TypeScript also
    uses short circuit evaluation, if the left side
    of && is already false the right side is never
    evaluated.

* **Example B.5 — Ternary:** [Example B.5](./project-overview/language-analysis/B-Expressions_Assignment-Statements/example_2-5.ts)
  * **Implementation:**
    ```typescript
    let score: number = 67;
    let result: string = score >= 75 ? "Passed" : "Failed";

    console.log(result);   // Failed
    ```
  * **Analysis:** The ternary operator works by checking a condition ? valueIfTrue : valueIfFalse.
    It is a shorthand if/else that fits in one line
    and returns a value directly. Best used for simple
    single condition checks. Nesting multiple ternary
    expressions makes code harder to read and should
    be avoided.

* **Example B.6 — Nullish Coalescing:** [Example B.6](./project-overview/language-analysis/B-Expressions_Assignment-Statements/example_2-6.ts)
  * **Implementation:**
    ```typescript
    let playerName: string | null = null;
    let displayName: string = playerName ?? "Guest";

    console.log(displayName);   // Guest
    ```
  * **Analysis:** The ?? nullish coalescing operator
    returns the right side only if the left side is null or
    undefined. This is different from JavaScript's ||
    operator which returns the right side for any
    falsy value including 0 and empty string. The ??
    operator is more precise because it only triggers
    on null and undefined making it safer for numeric
    and boolean values that might legitimately be 0
    or false.

#### C. Statement-Level Control Structures

* **Example C.1 — If/Else:** [Example C.1](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-1.ts)
  * **Implementation:**
    ```typescript
    let score: number = 67;

    if (score >= 90) {
        console.log("Excellent");
    } else if (score >= 75) {
        console.log("Passed");
    } else {
        console.log("Failed");
    }
    ```
  * **Analysis:** The if/else, else/if statement are
    conditions that are evaluated top
    to bottom and the first matching block executes.
    TypeScript inherits JavaScript's
    truthy/falsy system, any non-null, non-zero,
    non-empty value is truthy. TypeScript with strict
    mode enabled enforces that conditions in if
    statements must be boolean expressions reducing
    accidental truthy/falsy bugs.

* **Example C.2 — Switch:** [Example C.2](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-2.ts)
  * **Implementation:**
    ```typescript
    let day: number = 3;

    switch (day) {
        case 1:
            console.log("Monday");
            break;
        case 2:
            console.log("Tuesday");
            break;
        case 3:
            console.log("Wednesday");
            break;
        default:
            console.log("Other day");
            break;
    }
    ```
  * **Analysis:** The switch statement has
    cases that needs a break to prevent fall
    through. The default case runs if no match is
    found. TypeScript switch is very flexible
    because it can switch on strings, numbers,
    and even computed values without needing a special
    pattern matching syntax.

* **Example C.3 — Switch with String:** [Example C.3](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-3.ts)
  * **Implementation:**
    ```typescript
    let grade: string = "B";

    switch (grade) {
        case "A":
            console.log("Excellent");
            break;
        case "B":
            console.log("Good");
            break;
        case "C":
            console.log("Average");
            break;
        default:
            console.log("Invalid grade");
    }
    ```
  * **Analysis:** TypeScript switch works on string
    values directly without needing a special syntax. The string comparison
    uses strict equality === internally so the match
    must be exact. This is cleaner than writing
    multiple if/else if chains when checking one
    variable against many possible string values.

* **Example C.4 — For Loop:** [Example C.4](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-4.ts)
  * **Implementation:**
    ```typescript
    for (let i: number = 1; i <= 5; i++) {
        console.log(`Count: ${i}`);
    }
    ```
  * **Analysis:** The for loop has three parts,
    initialization, condition,and increment. 
    The loop runs as long as thecondition is true.
    TypeScript uses let for loop variables instead of var to ensure block scoping,
    a let variable declared inside the for loop is
    not accessible outside of it which prevents
    common JavaScript scoping bugs.

* **Example C.5 — For...of:** [Example C.5](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-5.ts)
  * **Implementation:**
    ```typescript
    let players: string[] = ["John Kaisen", "BOB", "Yuji Itadori"];

    for (let player of players) {
        console.log(player);
    }
    ```
  * **Analysis:** The for...of loop is TypeScript's
    forloop, It iterates over
    the values of an iterable like an array without
    needing an index. It is cleaner and safer than
    a regular for loop when you just need every
    element. TypeScript also has for...in which
    iterates over keys/indexes instead of values, 
    for...of is generally preferred for arrays.

* **Example C.6 — While:** [Example C.6](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-6.ts)
  * **Implementation:**
    ```typescript
    let count: number = 1;

    while (count <= 5) {
        console.log(`Count: ${count}`);
        count++;
    }
    ```
  * **Analysis:** The while loop checks the condition before each iteration, 
    if false from the start the loop body never runs.
    The count variable must be updated inside the loop
    otherwise it runs forever as an infinite loop.
    Used when the number of iterations is not known
    ahead of time.

* **Example C.7 — Do/While:** [Example C.7](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-7.ts)
  * **Implementation:**
    ```typescript
    let count: number = 1;

    do {
        console.log(`Count: ${count}`);
        count++;
    } while (count <= 5);
    ```
  * **Analysis:** The do/while loop checks the condition
    after the loop body runs guaranteeing the body
    executes at least once. Useful for showing menus
    or prompts where you always want to display
    something at least one time before checking if
    the user wants to continue.

* **Example C.8 — Break and Continue:** [Example C.8](./project-overview/language-analysis/C-Statement-Level_Control-Structures/example_3-8.ts)
  * **Implementation:**
    ```typescript
    for (let i: number = 1; i <= 10; i++) {
        if (i === 3) continue;   // skip 3
        if (i === 7) break;      // stop at 7

        console.log(i);
    }
    ```
  * **Analysis:** break and continue are guard
    statements that control loop flow. continue skips
    the rest of the current iteration and moves to
    the next one. break exits the loop entirely.
    They should be used sparingly, overusing them can make loop
    logic harder to follow.

#### D. Subprograms

* **Example D.1 — Basic Function:** [Example D.1](./project-overview/language-analysis/D-Subprograms/example_4-1.ts)
  * **Implementation:**
    ```typescript
    function add(a: number, b: number): number {
        return a + b;
    }

    console.log(add(5, 3));   // 8
    ```
  * **Analysis:** This is a basic typed function in
    TypeScript. Parameter types and return type are
    explicitly declared. TypeScript enforces that the
    function always returns a number, returning a
    string or nothing would throw a compile error.
    Parameters are passed by value meaning the function
    gets its own copy of the values and changes inside
    do not affect the original variables.

* **Example D.2 — Optional Parameters:** [Example D.2](./project-overview/language-analysis/D-Subprograms/example_4-2.ts)
  * **Implementation:**
    ```typescript
    function greet(name: string, greeting: string = "Hello"): void {
        console.log(`${greeting} ${name}`);
    }

    greet("BOB");             // Hello BOB
    greet("John Kaisen", "Welcome");    // Welcome John Kaisen
    ```
  * **Analysis:** TypeScript supports default parameter
    values. If the caller does not
    provide the argument the default value is used.
    Optional parameters must come after required
    parameters. The void return type means the function
    returns nothing.

* **Example D.3 — Recursion:** [Example D.3](./project-overview/language-analysis/D-Subprograms/example_4-3.ts)
  * **Implementation:**
    ```typescript
    function factorial(n: number): number {
        if (n <= 1) return 1;
        return n * factorial(n - 1);
    }

    console.log(factorial(5));   // 120
    ```
  * **Analysis:** Recursion in TypeScript works when the
    function calls itself with a
    smaller input until the base case n <= 1 is
    reached. Without the base case the function
    would call itself infinitely until a stack
    overflow occurs. TypeScript enforces the return
    type number so every code path must return a
    number or the compiler throws an error.

* **Example D.4 — Rest Parameters:** [Example D.4](./project-overview/language-analysis/D-Subprograms/example_4-4.ts)
  * **Implementation:**
    ```typescript
    function sumAll(...numbers: number[]): number {
        return numbers.reduce((total, n) => total + n, 0);
    }

    console.log(sumAll(1, 2, 3));          // 6
    console.log(sumAll(1, 2, 3, 4, 5));    // 15
    ```
  * **Analysis:** Rest parameters using ... allow a
    function to accept any number of arguments as an
    array. The ...numbers parameter collects
    all passed arguments into a typed number array.
    reduce then adds them all together. This is more
    flexible than defining a fixed number of parameters
    when the count of inputs is not known ahead of time.

* **Example D.5 — Arrow Function:** [Example D.5](./project-overview/language-analysis/D-Subprograms/example_4-5.ts)
  * **Implementation:**
    ```typescript
    const multiply = (a: number, b: number): number => a * b;
    const printMessage = (message: string): void => console.log(message);

    console.log(multiply(4, 5));          // 20
    printMessage("Hello from arrow");
    ```
  * **Analysis:** Arrow functions are TypeScript's
    concise function syntax. They are assigned to variables and
    can be passed around as values. Arrow functions
    also have a different this binding than regular
    functions which is important in class and callback
    contexts. For short single expression functions
    the curly braces and return keyword can be omitted
    making them very concise.

* **Example D.6 — Closure:** [Example D.6](./project-overview/language-analysis/D-Subprograms/example_4-6.ts)
  * **Implementation:**
    ```typescript
    function createMultiplier(factor: number): (n: number) => number {
        return (number: number) => number * factor;
    }

    const triple = createMultiplier(3);
    console.log(triple(5));    // 15
    console.log(triple(10));   // 30
    ```
  * **Analysis:** A closure is a function that captures
    and remembers variables from its enclosing scope
    even after that scope has finished executing. Here
    the returned arrow function captures factor from
    the outer function. Every time triple is called it
    still has access to factor even though
    createMultiplier has already returned. It is a powerful
    pattern for creating reusable parameterized
    functions without needing a full class.

* **Example D.7 — Named Parameters via Object:** [Example D.7](./project-overview/language-analysis/D-Subprograms/example_4-7.ts)
  * **Implementation:**
    ```typescript
    function printInfo({ name, age }: { name: string; age: number }): void {
        console.log(`${name} is ${age} years old`);
    }

    printInfo({ name: "John Kaisen", age: 25 });
    printInfo({ age: 30, name: "BOB" });
    ```
  * **Analysis:** TypeScript achieves named parameters
    through object destructuring in the function
    signature. The caller passes an object with named
    properties which can be in any order. This is
    very flexible because the
    object destructuring pattern also works with
    complex nested structures. The types are enforced
    inline in the destructuring pattern ensuring the
    compiler validates the shape of the passed object.

#### E. Abstract Data Types and Encapsulation

* **Example E.1 — Namespace as Module:** [Example E.1](./project-overview/language-analysis/E-Abstract-Data-Types_Encapsulation/example_5-1.ts)
  * **Implementation:**
    ```typescript
    namespace PlayerStats {
        let highScore: number = 0;

        export function setHighScore(score: number): void {
            if (score > highScore) highScore = score;
        }

        export function getHighScore(): number {
            return highScore;
        }
    }

    PlayerStats.setHighScore(200);
    PlayerStats.setHighScore(150);
    console.log(PlayerStats.getHighScore());   // 200
    ```
  * **Analysis:** TypeScript namespaces group related
    logic together and control what is exposed to the
    outside. Only functions marked with export are
    accessible from outside the namespace. highScore
    is not exported so it cannot be accessed or
    modified directly from outside, only through
    setHighScore and getHighScore which enforce their
    own rules.

* **Example E.2 — Data Hiding:** [Example E.2](./project-overview/language-analysis/E-Abstract-Data-Types_Encapsulation/example_5-2.ts)
  * **Implementation:**
    ```typescript
    namespace BankSystem {
        let balance: number = 1000;

        export function deposit(amount: number): void {
            if (amount > 0) balance += amount;
        }

        export function withdraw(amount: number): void {
            if (amount > 0 && amount <= balance) balance -= amount;
        }

        export function getBalance(): number {
            return balance;
        }
    }

    BankSystem.deposit(500);
    BankSystem.withdraw(200);
    console.log(BankSystem.getBalance());   // 1300
    ```
  * **Analysis:** balance is not exported meaning
    no outside code can directly set it to any
    arbitrary value. All interactions go through
    deposit, withdraw, and getBalance each of which
    enforces its own validation rules. This is data
    hiding in action, the internal state is protected
    and only controlled operations are exposed to
    the outside.

* **Example E.3 — Input Validation:** [Example E.3](./project-overview/language-analysis/E-Abstract-Data-Types_Encapsulation/example_5-3.ts)
  * **Implementation:**
    ```typescript
    namespace GameSettings {
        let volume: number = 50;
        let fullscreen: boolean = false;

        export function setVolume(value: number): void {
            if (value >= 0 && value <= 100) volume = value;
        }

        export function setFullscreen(value: boolean): void {
            fullscreen = value;
        }

        export function showSettings(): void {
            console.log(`Volume: ${volume}, Fullscreen: ${fullscreen}`);
        }
    }

    GameSettings.setVolume(80);
    GameSettings.setFullscreen(true);
    GameSettings.showSettings();   // Volume: 80, Fullscreen: true
    ```
  * **Analysis:** Both volume and fullscreen are
    hidden inside the namespace. setVolume enforces
    a valid range of 0 to 100, any value outside
    that range is silently ignored. This ensures
    the internal state always stays valid no matter
    what values outside code tries to pass in.

* **Example E.4 — Controlled State:** [Example E.4](./project-overview/language-analysis/E-Abstract-Data-Types_Encapsulation/example_5-4.ts)
  * **Implementation:**
    ```typescript
    namespace Counter {
        let count: number = 0;

        export const increment = (): void => { count++; };
        export const decrement = (): void => { if (count > 0) count--; };
        export const reset = (): void => { count = 0; };
        export const getCount = (): number => count;
    }

    Counter.increment();
    Counter.increment();
    Counter.increment();
    Counter.decrement();
    console.log(Counter.getCount());   // 2
    Counter.reset();
    console.log(Counter.getCount());   // 0
    ```
  * **Analysis:** count is fully hidden inside the
    namespace, outside code can never directly set
    it to any value. It can only go up through
    increment, down through decrement which prevents
    it from going below 0, or back to zero through
    reset. This guarantees the counter can never
    end up in an invalid state.

* **Example E.5 — Collection Hiding:** [Example E.5](./project-overview/language-analysis/E-Abstract-Data-Types_Encapsulation/example_5-5.ts)
  * **Implementation:**
    ```typescript
    namespace Inventory {
        let items: string[] = [];

        export function addItem(item: string): void {
            if (item && item.trim() !== "") items.push(item);
        }

        export function removeItem(item: string): void {
            items = items.filter(i => i !== item);
        }

        export function showItems(): void {
            items.forEach(item => console.log(item));
        }

        export function getCount(): number {
            return items.length;
        }
    }

    Inventory.addItem("Sword");
    Inventory.addItem("Shield");
    Inventory.addItem("");
    Inventory.showItems();
    console.log(Inventory.getCount());   // 2
    ```
  * **Analysis:** The items array is fully hidden, 
    outside code cannot directly add, remove, or
    clear it. All interactions go through addItem
    which rejects empty strings, removeItem, and
    showItems. The empty string passed in addItem("")
    gets silently rejected because the validation
    inside blocks it. This pattern of hiding a
    collection behind controlled methods is extremely
    common in real TypeScript applications.

#### F. Object-Oriented Programming

* **Example F.1 — Inheritance:** [Example F.1](./project-overview/language-analysis/F-Object-Oriented-Programming/example_6-1.ts)
  * **Implementation:**
    ```typescript
    class Animal {
        protected name: string;

        constructor(name: string) {
            this.name = name;
        }

        speak(): void {
            console.log(`${this.name} makes a sound`);
        }
    }

    class Dog extends Animal {
        constructor(name: string) {
            super(name);
        }

        speak(): void {
            console.log(`${this.name} says Woof`);
        }
    }

    const animal: Animal = new Dog("BOB");
    animal.speak();   // BOB says Woof
    ```
  * **Analysis:** Inheritance in TypeScript uses the
    extends keyword same as Java. The super() call
    passes arguments up to the parent constructor.
    The child class overrides the speak method with
    its own implementation, TypeScript
    does not require the virtual or override keywords.
    Methods are virtual by default in TypeScript
    meaning child classes can always override them
    without any special declaration.

    to be treated as the same base type while still
    behaving differently. All three objects are stored

    at runtime. TypeScript resolves the correct method

    not the declared type.

        describe(): void {
            console.log(`This shape has an area of ${this.getArea()}`);
        }

        }
    }

        constructor(private width: number, private height: number) { super(); }

            return this.width * this.height;
        }
    }

    const rectangle: Shape = new Rectangle(4, 6);

    console.log(rectangle.getArea());    // 24

  * **Analysis:** The abstract keyword prevents
    the class from being instantiated directly and
    forces child classes to implement all abstract
    methods. Non-abstract methods like describe are
    inherited automatically. TypeScript also supports
    the private constructor parameter shorthand, 
    private radius: number in the constructor
    automatically creates and assigns a private
    class field without needing a separate declaration.

#### G. Concurrency

* **Example G.1 — Async/Await:** [Example G.1](./project-overview/language-analysis/G-Concurrency/example_7-1.ts)
  * **Implementation:**
    ```typescript
    async function fetchData(): Promise<void> {
        console.log("Fetching data...");
        await new Promise(resolve => setTimeout(resolve, 2000));
        console.log("Data fetched");
    }

    fetchData();
    ```
  * **Analysis:** async and await use Promises instead of Tasks. An async
    function always returns a Promise. The await
    keyword pauses execution until the Promise
    resolves without blocking the event loop.
    TypeScript uses a single threaded event loop
    model inherited from JavaScript, 
    TypeScript/Node.js handles concurrency
    through non-blocking I/O and the event queue.

* **Example G.2 — Promise.all:** [Example G.2](./project-overview/language-analysis/G-Concurrency/example_7-2.ts)
  * **Implementation:**
    ```typescript
    async function printNumbers(label: string): Promise<void> {
        for (let i = 1; i <= 3; i++) {
            console.log(`${label}: ${i}`);
            await new Promise(resolve => setTimeout(resolve, 100));
        }
    }

    async function main(): Promise<void> {
        await Promise.all([
            printNumbers("Task 1"),
            printNumbers("Task 2")
        ]);

        console.log("Both tasks done");
    }

    main();
    ```
  * **Analysis:** Promise.all runs multiple
    async operations concurrently and waits for all
    of them to complete. Both tasks run at the same
    time and their output interleaves. Node.js
    is still single threaded and the interleaving
    happens through the event loop not actual
    simultaneous execution.

* **Example G.3 — Promise Chain:** [Example G.3](./project-overview/language-analysis/G-Concurrency/example_7-3.ts)
  * **Implementation:**
    ```typescript
    async function getScore(): Promise<number> {
        return new Promise(resolve => {
            setTimeout(() => resolve(100), 3000);
        });
    }

    async function main(): Promise<void> {
        const result = await getScore();
        console.log(`Score: ${result}`);   // Score: 100
    }

    main();
    ```
  * **Analysis:** A Promise represents an eventual
    value, it will either resolve with a value or
    reject with an error. The await unwraps
    the Promise and gives back the actual resolved
    value. setTimeout simulates an async operation
    like an API call. getScore returns Promise<number>
    meaning it will eventually produce a number
    once the async work completes.

* **Example G.4 — Try/Catch with Async:** [Example G.4](./project-overview/language-analysis/G-Concurrency/example_7-4.ts)
  * **Implementation:**
    ```typescript
    async function riskyOperation(): Promise<string> {
        throw new Error("Something went wrong");
    }

    async function main(): Promise<void> {
        try {
            const result = await riskyOperation();
            console.log(result);
        } catch (error) {
            console.log(`Caught: ${(error as Error).message}`);
        }
    }

    main();
    ```
  * **Analysis:** Errors in async functions are
    handled using try/catch around the await call.
    When an async function throws an error the
    Promise rejects and the catch block catches
    the rejection. Theerror as Error cast is needed because TypeScript
    types caught errors as unknown by default in
    strict mode.

* **Example G.5 — Parallel Execution:** [Example G.5](./project-overview/language-analysis/G-Concurrency/example_7-5.ts)
  * **Implementation:**
    ```typescript
    async function processItem(id: number): Promise<string> {
        await new Promise(resolve => setTimeout(resolve, 100));
        return `Item ${id} processed`;
    }

    async function main(): Promise<void> {
        const results = await Promise.all(
            [0, 1, 2, 3, 4].map(i => processItem(i))
        );

        results.forEach(r => console.log(r));
        console.log("All items processed");
    }

    main();
    ```
  * **Analysis:** Promise.all with a mapped array
    launches all async operations concurrently and
    collects all results into an array once they all
    complete. This still runs on a single
    thread but uses non-blocking I/O to handle
    multiple operations at the same time through
    the event loop.

* **Example G.6 — Async Return Value:** [Example G.6](./project-overview/language-analysis/G-Concurrency/example_7-6.ts)
  * **Implementation:**
    ```typescript
    async function getUserName(id: number): Promise<string> {
        await new Promise(resolve => setTimeout(resolve, 500));

    async function main(): Promise<void> {

    }

    ```
  * **Analysis:** Promise<string> is used when an
    async function needs to return a value,
    The await unwraps the Promise and gives back

#### H. Exception and Event Handling

* **Example H.1 — Try/Catch/Finally:** [Example H.1](./project-overview/language-analysis/H-Exception_Event-Handling/example_8-1.ts)
  * **Implementation:**
    ```typescript
    try {
        const numbers: number[] = [1, 2, 3];
        const value = numbers[5];
        if (value === undefined) throw new Error("Index out of bounds");
        console.log(value);
    } catch (error) {
        console.log(`Error: ${(error as Error).message}`);
    } finally {
        console.log("This always runs");
    }
    ```
  * **Analysis:** The try block contains code that might
    throw. The catch block handles the error
    gracefully. The finally block always runs
    regardless. accessing
    an out of bounds array index returns undefined
    silently. So the check must be done manually.
    The error as Error cast is needed because
    TypeScript types caught errors as unknown in
    strict mode.

* **Example H.2 — Multiple Catch Types:** [Example H.2](./project-overview/language-analysis/H-Exception_Event-Handling/example_8-2.ts)
  * **Implementation:**
    ```typescript
    function parseInput(input: string | null): number {
        if (input === null) throw new TypeError("Input cannot be null");
        const result = Number(input);
        if (isNaN(result)) throw new RangeError("Input is not a valid number");
        return result;
    }

    try {
        console.log(parseInput(null));
    } catch (error) {
        if (error instanceof TypeError) {
            console.log(`Type Error: ${error.message}`);
        } else if (error instanceof RangeError) {
            console.log(`Range Error: ${error.message}`);
        } else {
            console.log(`General Error: ${(error as Error).message}`);
        }
    }
    ```
  * **Analysis:** Different error
    types are handled using instanceof checks inside
    a single catch block. This achieves the same
    result but requires more if/else logic inside
    the catch. TypeError and RangeError are built in
    JavaScript error types.

* **Example H.3 — Throw:** [Example H.3](./project-overview/language-analysis/H-Exception_Event-Handling/example_8-3.ts)
  * **Implementation:**
    ```typescript
    function setAge(age: number): void {
        if (age < 0) throw new Error("Age cannot be negative");
        console.log(`Age set to ${age}`);
    }

    try {
        setAge(-5);
    } catch (error) {
        console.log(`Error: ${(error as Error).message}`);
    }
    ```
  * **Analysis:** The throw keyword manually triggers
    an error when a condition is violated. This is
    defensive programming, instead of silently
    accepting bad data the function immediately
    signals that something invalid was passed.
    The caller handles it in their own try/catch
    block.

* **Example H.4 — Custom Exception:** [Example H.4](./project-overview/language-analysis/H-Exception_Event-Handling/example_8-4.ts)
  * **Implementation:**
    ```typescript
    class InvalidHealthError extends Error {
        constructor(message: string) {
            super(message);
            this.name = "InvalidHealthError";
        }
    }

    function validateHealth(health: number): void {
        if (health < 0) throw new InvalidHealthError("Health cannot be below zero");
    }

    try {
        validateHealth(-10);
    } catch (error) {
        if (error instanceof InvalidHealthError) {
            console.log(`Custom Error: ${error.message}`);
        }
    }
    ```
  * **Analysis:** Custom exceptions are created by
    extending the Error class. Setting this.name gives the error
    a meaningful name that shows up in stack traces.
    Custom exceptions make error handling more
    readable and allow callers to catch specific
    application errors separately from built in
    system errors using instanceof checks.

* **Example H.5 — Event Handling:** [Example H.5](./project-overview/language-analysis/H-Exception_Event-Handling/example_8-5.ts)
  * **Implementation:**
    ```typescript
    type EventHandler = () => void;

    class Player {
        private health: number = 100;
        private deathHandlers: EventHandler[] = [];

        onDeath(handler: EventHandler): void {
            this.deathHandlers.push(handler);
        }

        takeDamage(amount: number): void {
            this.health -= amount;
            console.log(`Player took ${amount} damage`);
            if (this.health <= 0) {
                this.deathHandlers.forEach(h => h());
            }
        }
    }

    const player = new Player();
    player.onDeath(() => console.log("Player has died!"));
    player.onDeath(() => console.log("Respawning player..."));
    player.takeDamage(100);
    // Player took 100 damage
    // Player has died!
    // Respawning player...
    ```
  * **Analysis:** TypeScript implements the event
    pattern using arrays of callback functions.
    TypeScript uses function
    types and arrays to store multiple handlers.
    The EventHandler type alias defines what a
    valid handler looks like. When the event fires
    all registered handlers are called using
    forEach.

* **Example H.6 — Error Propagation:** [Example H.6](./project-overview/language-analysis/H-Exception_Event-Handling/example_8-6.ts)
  * **Implementation:**
    ```typescript
    function readConfig(): string {
        throw new Error("Config file not found");
    }

    function loadApp(): void {
        try {
            const config = readConfig();
            console.log(config);
        } catch (error) {
            console.log(`App failed to load: ${(error as Error).message}`);
            throw error;
        }
    }

    try {
        loadApp();
    } catch (error) {
        console.log(`Top level caught: ${(error as Error).message}`);
    }
    ```
  * **Analysis:** Error propagation shows how errors
    bubble up through multiple function call levels.
    loadApp catches the error from readConfig, logs
    it, and then re-throws it using throw error so
    the caller also gets a chance to handle it.
    It is useful when a function needs to log or partially
    handle an error but still wants the outer caller
    to know something went wrong.

## 3. Comparative Analysis Summary

| **Concept** | **Language A: TypeScript** | **Language B: PHP** | **Comparative Conclusion** |
| :--- | :--- | :--- | :--- |
| **Data Types** | Statically typed with type annotations, number, string, boolean, any, unknown. Type inference available. Types checked at compile time. | Dynamically typed, variables have no declared type and can change type freely at runtime. Type hints available in PHP 7+ but not strictly enforced. | TypeScript catches type errors at compile time before the program runs making it safer for large codebases. PHP's dynamic typing is faster to write but type errors only show up at runtime which makes debugging harder. |
| **Expressions and Assignment** | Supports compound assignment, ternary, and nullish coalescing ?? which only triggers on null and undefined. Uses === for strict equality. | Supports compound assignment, ternary, and null coalescing ??. Uses == with type coercion and === for strict equality. | TypeScript's === default behavior and strict null checking makes comparisons more predictable. PHP's == performs type coercion which can cause unexpected results like "1" == 1 being true. |
| **Control Structures** | Standard if/else, switch, for, for...of, while, do/while, break/continue. | Standard if/else, switch, for, foreach, while, do/while, break/continue. PHP has a unique foreach specifically for arrays and associative arrays. | Both are similar. PHP's foreach is slightly more convenient for associative arrays. TypeScript's for...of is cleaner for iterating typed collections. |
| **Subprograms** | Functions are typed with parameter and return types. Supports arrow functions, closures, rest parameters, and default parameters. | Functions are dynamically typed. Supports default parameters, variadic functions with ...$args, closures, and anonymous functions. | TypeScript function signatures are more explicit making it clearer what a function expects and returns. PHP functions are faster to write but harder to understand without documentation since types are optional. |
| **Encapsulation** | Achieved through namespaces with export keyword controlling visibility. Class access modifiers private, public, protected enforced at compile time. | Achieved through classes with private, public, protected access modifiers. Namespaces exist but are rarely used, classes are the primary encapsulation unit. PHP access modifiers are enforced at runtime. | TypeScript encapsulation is enforced at compile time catching violations before the code runs. PHP enforces at runtime so violations only appear when that code actually executes. |

Methods are virtual by default. | Fully OOP, supports inheritance with extends, abstract classes, interfaces with implements, traits for code reuse, and access modifiers. No method overloading, uses default parameters instead. | TypeScript OOP is stricter with compile time checking. ### Overall Conclusion

TypeScript is the stronger language for large scale frontend and
full stack applications where type safety, tooling, and
maintainability matter. Its compile time type checking, mature
async model, and strong OOP support make it reliable for complex
systems.

PHP remains strong for server side web development especially
with frameworks like Laravel. It is faster to get running,
has simpler concurrency for typical web request/response
workflows, and its multiple catch syntax is cleaner. However
its dynamic typing and runtime-only enforcement make it harder
to maintain at scale compared to TypeScript.

## 4. "Code Smell" & Refactoring Reflection

**Original Code Link:** [smelly-code.ts](https://github.com/ccsad-aquamarine/Language-Implementation-Audit/blob/main/proj-overview/code-smell/smelly-code.ts)

**The Smell:** The smell is creating unnecessary object literals inside a 
loop when the object is only used within that iteration and 
never passed or stored. Every new {} allocates heap memory 
and adds pressure to the garbage collector. The fix is to 
lift the variables outside the loop and reuse them, 
primitives are stack allocated and have zero garbage 
collection overhead.
#
**The Refactor:** (Use the code cell below)

````typescript
// REFACTORED — Clean Implementation
// Reusing variables instead of creating new objects
// Faster execution and significantly lower memory usage

export {};

const start = performance.now();
const before = process.memoryUsage().heapUsed;

let total: number = 0;
let x: number = 0;
let y: number = 0;
let z: number = 0;

for (let i = 0; i < 100000; i++) {
    x = i;
    y = i * 2;
    z = i * 3;
    total += x + y + z;
}

const end = performance.now();
const after = process.memoryUsage().heapUsed;

console.log("=== Variable Reuse ===");
console.log(`Total: ${total}`);
console.log(`Time: ${(end - start).toFixed(3)}ms`);
console.log(`Memory: ${after - before} bytes`);
````

## 5. Final Reflection Questions

1. **Paradigm & Language Shift:** If you had to rebuild this project in a significantly different modern programming language (e.g., Rust, Go, Swift, or Elixir), which of the 8 concepts would be the most difficult to translate or would require the most significant architectural change? Why?

- If we had to rebuild this project in Rust, the most difficult
concept to translate would be Concurrency. In TypeScript
concurrency is handled through Promises and async/await on a
single threaded event loop, you never have to think about
who owns the data or whether two threads are accessing the
same variable at the same time because there is only ever
one thread running at once. Rust does not have a garbage
collector or an event loop. Instead it uses an ownership and
borrowing system where the compiler itself enforces memory
safety at compile time. Every concurrent operation has to be
explicitly designed around who owns the data, you cannot
just await something and move on. Encapsulation would also
require significant rethinking since Rust uses a module
system with pub and private visibility that works differently
from TypeScript's namespace and class based access modifiers.
The architectural change would not just be a rewrite, it
would be a completely different way of thinking about how
data moves through a program.


2. **Language Evaluation:** Based on this audit, did the language(s) you chose help or hinder the development of your software engineering project? (Cite specific concepts like Concurrency or Exception Handling).

- TypeScript genuinely helped the development of our Cliftunes
project. Exception Handling was one of the most useful
concepts, the project involves a lot of runtime operations
like fetching data, processing user inputs, and handling
edge cases where values might be null or undefined. Having
structured try/catch blocks with custom error types meant
we could handle those failures gracefully without the app
breaking entirely. Concurrency through async/await was the
backbone of the project, almost every operation that touches

number of bugs at compile time that would have only shown

felt like the right tool for the project.
